```python
# import all necessary packages.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('sales_data (4).csv',encoding='latin1')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
## checking duplicate data
df.duplicated().sum()
```

```python
# droping all the duplicate rows
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.head()
```

```python
# checking null values in status and unnmaed 1
df['Status'].isnull().sum()
df['unnamed1'].isnull().sum()
```

```python
# Dropping unwanted columns.
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
df.columns
```

```python
df.info()
```

```python
# checking the null values present in Amount.
df['Amount'].isna().sum()
```

```python
# deleting null rows
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
# Changing the Values
gen = {'F':'Female','M':'Male'}
df['Gender'] = df['Gender'].map(gen)
```

```python
df.head()
```

```python
# changing the datatype of Marital_Status
df['Marital_Status'] = df['Marital_Status'].astype(str)
```

```python
df['Marital_Status'] = df['Marital_Status'].str.replace('0','Single')
df['Marital_Status'] = df['Marital_Status'].str.replace('1','Married')
```

```python
df.head()
```

```python
df['State'].unique()
```

```python
df['State'] = df['State'].str.replace('\xa0',' ')
```

```python
df['State'].unique()
```

```python
df.head()
```

```python
def label(a):
  for i in a.containers:
    a.bar_label(i)
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(20,10))
c_g = sns.countplot(x='Gender',data=df,hue='Gender',ax=axes[0])
label(c_g)
gr_g = df.groupby('Gender')['Amount'].sum()
gr_g = pd.DataFrame(gr_g)
gr_g.reset_index(inplace=True)
axes[1].pie(gr_g['Amount'],labels=gr_g['Gender'].unique(),autopct='%0.0f%%',colors=['#FFA1AD','#FF6467'])
plt.show()
```

Females are more spending on Shopping

```python
# who spend more on shoppin by age group
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(15,7))
age_g = df.groupby('Age Group')['Amount'].sum()
age_g = pd.DataFrame(age_g)
age_g.reset_index(inplace=True)

ab= sns.barplot(x='Age Group',y='Amount',ax=axes[0],data=age_g.loc[0:4],hue='Age Group')
label(ab)
#####
state_gr = df.groupby('State')['Amount'].sum()
state_gr = pd.DataFrame(state_gr)
state_gr.sort_values(by='Amount',ascending=False,inplace=True)
state_gr.reset_index(inplace=True)
# state_gr.drop('index',axis=1,inplace=True)
b = sns.barplot(x='State',y='Amount',data=state_gr.loc[0:4],hue='State'
                ,palette='coolwarm',ax=axes[1])
label(b)

```

The age group between 26-35 is spending more money on shopping

```python
plt.figure(figsize=(20,10))
sta = sns.barplot(x='Occupation',y='Amount',data=df,hue='Occupation'
                  ,errorbar=None,estimator=sum)
label(sta)
```

**Conclusion: Based on the above analysis I can conclude that Females of 26-35 from Uttar-Pradesh,Maharashtra and Karnataka working IT,Healthcare and Aviation spend more money on Food,clothes and footware and Electrocin gadgets.**
